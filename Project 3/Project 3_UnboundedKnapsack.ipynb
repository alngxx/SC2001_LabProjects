```python
def unbounded_knapsack(C, n, p, w):
    # Initialize 1D DP array 
    P = [0] * (C + 1)      # P[i] = max profit can obtained using knapsack capacity of i

    # Iterate through every knapsack capacity
    for c in range(1, C + 1):
        # For each capacity, try all object types
        for i in range(n):
            # If current object fit into capacity c
            if w[i] <= c:
                # Take the max of two choices:
                # 1. Skip object i: keep current best value stored in P[c]
                # 2. Choose object i: Add its profit to P[c - w[i]]
                P[c] = max(P[c], P[c - w[i]] + p[i])

    return P[C]
```

```python
# (4a): weights = [4, 6, 8], profits = [7, 6, 9]

C = 14
weights = [4, 6, 8]     
profits = [7, 6, 9]     
n = len(weights)

profit_4a = unbounded_knapsack(C, n, profits, weights)
print("4a: P(14) =", profit_4a)
```

```python
# (4b): weights = [5, 6, 8], profits = [7, 6, 9]

C = 14
weights = [5, 6, 8]     
profits = [7, 6, 9]     
n = len(weights)

profit_4b = unbounded_knapsack(C, n, profits, weights)
print("4b: P(14) =", profit_4b)
```
